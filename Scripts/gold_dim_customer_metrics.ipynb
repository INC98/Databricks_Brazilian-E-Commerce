```python
import pyspark.sql.functions as F
from pyspark.sql.window import Window
```

```python
query = """
SELECT
    customer_id,
    COUNT(DISTINCT order_id) AS total_orders,
    SUM(total_item_value) AS total_spent,
    SUM(total_item_value) / COUNT(DISTINCT order_id) AS avg_ticket

FROM `catalog_brazilian-e-commerce`.gold.fact_sales

GROUP BY customer_id
"""
df = spark.sql(query)
```

```python
df.limit(10).display()
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("`catalog_brazilian-e-commerce`.gold.dim_customer_metrics")
```
