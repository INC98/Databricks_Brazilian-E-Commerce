```python
import pyspark.sql.functions as F
from pyspark.sql.window import Window
```

```python
query = """SELECT
    f.seller_id,
    SUM(f.total_item_value) AS total_revenue,
    COUNT(DISTINCT f.order_id) AS total_orders,
    AVG(r.review_score) AS avg_review_score

FROM `catalog_brazilian-e-commerce`.gold.fact_sales f

LEFT JOIN `catalog_brazilian-e-commerce`.silver.olist_order_reviews r
    ON f.order_id = r.order_id

GROUP BY f.seller_id"""

df = spark.sql(query)
```

```python
df.limit(10).display()
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("`catalog_brazilian-e-commerce`.gold.dim_seller_performance")
```
