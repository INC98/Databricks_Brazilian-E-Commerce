```python
import pyspark.sql.functions as F
from pyspark.sql.window import Window
```

```python
query = """
SELECT
    p.product_category_name,
    SUM(f.total_item_value) AS total_revenue,
    SUM(f.freight_value) AS total_freight,
    COUNT(*) AS total_items_sold

FROM `catalog_brazilian-e-commerce`.gold.fact_sales f
JOIN `catalog_brazilian-e-commerce`.silver.olist_products p
    ON f.product_id = p.product_id

GROUP BY p.product_category_name
"""

df = spark.sql(query)
```

```python
df.limit(10).display()
```

# Crear gold table

```python
df.write.mode("overwrite").format("delta").saveAsTable("`catalog_brazilian-e-commerce`.gold.dim_sales_by_category")
```
